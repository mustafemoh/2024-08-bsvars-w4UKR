## 2. Are the libraries we need available?

```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import pandas as pd

print("numpy      :", np.__version__)
print("pandas     :", pd.__version__)
print("matplotlib :", matplotlib.__version__)
```

## 3. Can we do some linear algebra?

The models in this workshop are mostly matrix operations and loops, so let's check those.

```python
A = np.array([[2.0, 1.0],
              [1.0, 3.0]])
b = np.array([1.0, 2.0])

x = np.linalg.solve(A, b)   # solves A x = b
print("Solution x =", x)
print("Check A @ x =", A @ x, " (should be equal to b =", b, ")")
```

## 4. Can we make a figure?

```python
t = np.linspace(0, 4 * np.pi, 400)

plt.figure(figsize=(6, 3))
plt.plot(t, np.sin(t))
plt.title("If you can see this wave, figures work")
plt.xlabel("t")
plt.ylabel("sin(t)")
plt.tight_layout()
plt.show()
```
